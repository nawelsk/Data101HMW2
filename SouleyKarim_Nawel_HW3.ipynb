```python
#PART 2

ds = pd.read_csv('titanic.csv')
ds['Embarked'] = ds['Embarked'].fillna(ds['Embarked'].mode()[0])

#A
ds['Age'] = ds['Age'].fillna(ds['Age'].mean())

ds.head(25)
```

```python
#PART 3

ds = pd.read_csv('titanic.csv')
ds['FamilySize'] = ds['SibSp']+ ds['Parch'] + 1 
print(ds['FamilySize'].agg(['min','max','mean']))
FamilySize_1 = (ds['FamilySize'] == 1).sum()
print(FamilySize_1)
ds
```

```python
#PART 4

ds = pd.read_csv('titanic.csv')
female = ds[ds['Sex']=='female']

male = ds[ds['Sex']=='male']

under16 = ds[ds['Age']<16]

female['Survived'].mean() * 100
male['Survived'].mean() * 100
under16['Survived'].mean() * 100

df = {
'Group': ['Female', 'Male', 'Under 16'],
'Total': [
        female['PassengerId'].count(),      #Or len(...)
        male['PassengerId'].count(),
        under16['PassengerId'].count()],
'Survival Rate': [
        female['Survived'].mean() * 100,
        male['Survived'].mean() * 100,
        under16['Survived'].mean() * 100]
    }
df2 = pd.DataFrame(df)
print(df2)
```

```python
survival = ds.groupby(['Sex','Pclass'])['Survived'].mean().unstack()
survival
#unstack turn the second element in grouby into a column intead of only having rows
```

First of all, what we can observe is that female have the highest survival rate, meaning that they survived the most.
Second, the survival rate also depended on a second variable, which is the the passenger class. Passengers in first 
class survived more than those in the lower class. However, even with this condition, women still have highest survival rate.
