```python
import numpy as np

names = []
grades = np.zeros((10, 5))

for i in range(10):
    name = input("Enter student name: ")
    names.append(name)

    homework = float(input("Enter Homework grade: "))
    project = float(input("Enter Project grade: "))
    exam1 = float(input("Enter Exam 1 grade: "))
    exam2 = float(input("Enter Exam 2 grade: "))
    final_exam = float(input("Enter Final Exam grade: "))

    grades[i] = [homework, project, exam1, exam2, final_exam]    

weights = np.array([0.15, 0.15, 0.20, 0.20, 0.30])

final_grades = grades @ weights

letter_grades = []

for grade in final_grades:
    if grade >= 90:
        letter = "A"
    elif grade >= 80:
        letter = "B"
    elif grade >= 70:
        letter = "C"
    elif grade >= 60:
        letter = "D"
    else:
        letter = "F" 
    letter_grades.append(letter)
    
print("Final Grades:")

for i in range(10):
    print(names[i], ":", round(final_grades[i], 2), "-", letter_grades[i])
```
